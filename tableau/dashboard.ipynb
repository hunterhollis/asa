```python
import pandas as pd
import plotly.express as px
```

```python
df = px.data.gapminder().query("year==2007")
```

```python
df
```

```python
games_df = pd.read_csv('games_counts_data.csv')
games_df
```

```python
# find highest five day total for each metro
agg_df = games_df.copy()
agg_df['max_five_day_total'] = agg_df.groupby(by='metro_area')['five_day_total'].transform('max')
agg_df = agg_df[['metro_area', 'five_day_total', 'five_day_start_date', 'five_day_end_date', 'max_five_day_total']]
agg_df = agg_df[agg_df['max_five_day_total']==agg_df['five_day_total']].drop_duplicates().reset_index(drop=True)
agg_df
```

```python
agg_df_cities = agg_df[['metro_area', 'max_five_day_total']].drop_duplicates(keep='first').reset_index(drop=True)
agg_df_cities
```

```python
from geopy.geocoders import Nominatim
from geopy.extra.rate_limiter import RateLimiter
from tqdm import tqdm
tqdm.pandas()
geolocator = Nominatim(user_agent='na_cities')
geocode = RateLimiter(geolocator.geocode, min_delay_seconds=1)
```

```python
geocode("san antonio, texas")
```

```python
agg_df_loc = agg_df_cities.copy()
agg_df_loc['loc'] = agg_df_loc['metro_area'].progress_apply(geocode)
agg_df_loc
```

```python
agg_df_loc['latitude'] = agg_df_loc['loc'].apply(lambda x: x.latitude)
agg_df_loc['longitude'] = agg_df_loc['loc'].apply(lambda x: x.longitude)
agg_df_loc
```

```python
fig = px.scatter_geo(agg_df_loc, hover_name='metro_area', lat='latitude', lon='longitude', color='max_five_day_total', size='max_five_day_total', scope='north america', size_max=40)
# Use a conic projection and wider bounds so Canada is included while still highlighting subunit borders
fig.update_geos(center={'lat': 50, 'lon': -100}, lataxis_range=[25, 75], lonaxis_range=[-150, -50], showcountries=True, showcoastlines=True, showsubunits=True, subunitcolor='black', subunitwidth=1.5, coastlinecolor='black', coastlinewidth=0.8, projection_type='conic conformal')
# Increase the overall figure size and add a title
fig.update_layout(width=1200, height=800, title='Max Five Day Total by Metro Area (US + Canada)')
fig.show()
```
